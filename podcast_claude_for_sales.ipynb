## SEGMENT 2: Prospecting & Research — 7:00–14:00

**HOST:**
The first place Claude supercharges sales is prospecting — specifically, account research and building an ICP.

### ICP Development

If you haven't nailed your Ideal Customer Profile, Claude can help you build one from scratch or sharpen an existing one.

**Prompt:**
```
I sell [product/service]. My best customers tend to be [describe 2–3 current customers].
Help me define a detailed ICP including industry, company size, job titles, pain points,
and buying triggers.
```

### Account Research

Before any outreach or discovery call, you need context. Claude can compress hours of research into minutes.

**Prompt:**
```
Here's the LinkedIn profile and company about page for [prospect name] at [company].
What are their likely business challenges? What industry pressures is their company facing?
How might my solution — which does [X] — be relevant to them?
```

### Building Prospect Lists

**Prompt:**
```
I'm targeting VP of Operations at manufacturing companies with 200–500 employees in the Midwest.
What Boolean search string should I use in LinkedIn Sales Navigator?
What job titles might also hold this role?
```

## SEGMENT 3: Cold Outreach That Actually Gets Replies — 14:00–22:00

**HOST:**
Most cold outreach fails because it's generic. It talks about the seller, not the buyer. Claude helps you flip that.

### The Research-First Email (Framework: Research → Relevance → Request)

**Step 1 — Identify the angle:**
```
Here is the LinkedIn profile of [name], their company's recent press release about expanding
into Europe, and their job posting for a Head of Revenue Operations.
What are the 2–3 most likely pain points I should address in a cold email?
```

**Step 2 — Write the email:**
```
Write a cold email from me — [your name], [your title] at [your company] — that leads with
the European expansion challenge, connects it to how we help companies scale their sales
operations internationally, and ends with a low-friction CTA.
```

### Subject Line Testing

**Prompt:**
```
Generate 10 subject lines for this email. Include:
- 2 curiosity-based
- 2 that lead with the outcome
- 2 that reference [trigger event]
- 2 that ask a question
- 2 that are ultra-short (under 5 words)
```

### LinkedIn Outreach

**Prompt:**
```
Write a LinkedIn connection request message for [prospect]. Reference their post about [topic].
Keep it under 300 characters and don't pitch anything.
```

## SEGMENT 4: Discovery Call Prep — 22:00–28:00

**HOST:**
Discovery is arguably the most important part of the sales process, and preparation pays massive dividends. Claude can be your pre-call coach.

### Call Prep Brief

**Prompt:**
```
I have a discovery call tomorrow with [name], [title] at [company]. Here's what I know:
[paste LinkedIn, company info, previous email thread, notes].

Generate a call prep brief including:
- Their likely business priorities
- Key pain points to explore
- 5 discovery questions tailored to their situation
- Potential objections and how to handle them
- How our product is most likely relevant to them
```

### Discovery Question Bank

**Prompt:**
```
Generate 20 discovery questions for a VP of Sales at a mid-market B2B SaaS company.
Focus on: current sales process, pipeline visibility challenges, rep onboarding friction,
and quota attainment.
```

### Objection Prep

**Prompt:**
```
What are the 5 most common objections a [prospect type] would raise when evaluating
[your type of solution]? For each one, give me a reframe and a follow-up question
I can use to understand the objection more deeply before responding.
```

## SEGMENT 5: Proposals, Follow-Ups & Closing — 28:00–36:00

**HOST:**
Now we're in the deal — the prospect is engaged, you've had great discovery, and it's time to move toward a decision.

### Writing Proposals

**Prompt:**
```
Based on this discovery call transcript [or notes], write a proposal narrative for [company].
Lead with their primary pain around [X], explain how we solve it with [Y], quantify the
expected impact using [their numbers], and close with a clear recommended next step.
```

### Follow-Up Emails

**After a call:**
```
Write a follow-up email for a discovery call I just had with [name].
Key takeaways: [3 bullet points]. Next step: [demo in two weeks].
Recap the pain points in their language, confirm the next step,
and attach value by mentioning [relevant case study or insight].
```

**After sending a proposal:**
```
Write a gentle check-in email, 5 days after sending a proposal to [name].
Don't be pushy. Offer a useful insight about [their industry trend]
and reopen the conversation naturally.
```

### Handling Objections in Writing

**Prompt:**
```
A prospect just replied saying 'We don't have budget right now.'
Write a response that acknowledges this honestly, asks a clarifying question
to understand the timeline, and keeps the door open without being desperate.
```

### Closing Language

**Prompt:**
```
Write a closing email for a deal where we've had 3 meetings, they're aligned on value,
and I need to drive a decision this week due to end-of-quarter pricing.
Strike a balance between urgency and integrity — don't be manipulative.
```

## SEGMENT 6: Sales Enablement & Team Use Cases — 36:00–41:00

**HOST:**
If you're a sales leader, Claude isn't just a tool for individual reps — it's a force multiplier for your entire team.

### Building Playbooks
```
I'm going to describe how our best reps handle the discovery process. [Describe it.]
Turn this into a structured playbook with sections for: pre-call prep, opening the call,
key questions, transitioning to demo, and next steps.
```

### Buyer Persona Roleplay (Training)
```
Act as a skeptical VP of Finance at a 300-person manufacturing company evaluating a new
ERP system. I'm going to pitch you. Push back on pricing, ask hard ROI questions,
and bring up a competitor. Stay in character.
```

### Call Scoring Frameworks
```
Based on MEDDIC, create a call scoring rubric I can use to evaluate recorded discovery calls.
Include what 'good' looks like for each element and what red flags to watch for.
```

### Win/Loss Analysis
```
Here are notes from 5 deals we won and 3 we lost this quarter.
Identify patterns in the wins, patterns in the losses, and give me 3 process improvements
our team should consider.
```

## SEGMENT 7: Tips for Getting the Most Out of Claude — 41:00–44:00

**HOST:**
Five principles for using Claude effectively in sales:

| # | Principle | What It Means |
|---|---|---|
| 1 | **Context is everything** | Paste in LinkedIn profiles, email threads, company pages. More context = better output. |
| 2 | **Be specific about constraints** | Specify word count, tone, format, and audience. Vague prompts give generic results. |
| 3 | **Iterate, don't accept the first draft** | Say "make it shorter", "more casual", "rewrite the opening". Claude improves with feedback. |
| 4 | **Use it to think, not just to write** | Ask Claude to analyze, challenge, and stress-test. "What are the weaknesses in this proposal?" |
| 5 | **Keep your voice** | Edit the output so it sounds like you. Your authentic voice is your biggest competitive advantage. |
